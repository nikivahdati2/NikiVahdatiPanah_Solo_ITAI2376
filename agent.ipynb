```python
def search_notes(question):
    q = question.lower()

    if "neural network" in q:
        return "Neural Network: A system inspired by the human brain that learns patterns from data."

    elif "machine learning" in q:
        return "Machine Learning: A method where computers learn from data instead of being explicitly programmed."

    elif "deep learning" in q:
        return "Deep Learning: A type of machine learning that uses neural networks with many layers."

    elif "transformer" in q:
        return "Transformer: A deep learning model used in NLP that understands relationships between words."

    elif "overfitting" in q:
        return "Overfitting: When a model learns too much from training data and performs poorly on new data."

    else:
        return None
```

```python
def simple_explainer(text):
    return "In simple words: " + text
```

```python
def study_assistant(question):
    print("🧠 Step 1: Understanding the question...")

    print("🔧 Step 2: Calling tool 1 (search_notes)...")
    result = search_notes(question)

    if result:
        print("🔧 Step 3: Calling tool 2 (simple_explainer)...")
        answer = simple_explainer(result)
    else:
        print("🤔 Step 3: No result found")
        answer = "I couldn't find that in my notes."

    print("💾 Step 4: Saving conversation to memory...")
    remember_interaction(question, answer)

    return answer
```

```python
from langchain_core.tools import Tool

search_notes_tool = Tool(
    name="Search Study Notes",
    func=search_notes,
    description="Searches the AI study notes for beginner-friendly definitions."
)

simple_explainer_tool = Tool(
    name="Simple Explainer",
    func=simple_explainer,
    description="Turns information into simple beginner-friendly language."
)

tools = [search_notes_tool, simple_explainer_tool]

print("LangChain tools created successfully!")
```
